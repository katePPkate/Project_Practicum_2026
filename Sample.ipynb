```python
statistics = pd.read_parquet("customer_details/Показатели по регионам.parquet")

# Просмотр первых строк
display(statistics.head())
```

Оставляем только данные по региону, актуальному году, удаляю лишние столбцы

Создание id переменных для удобства 

**<font color="red">В DataFrame 'mapping' хранится инфа о переменных по id</font>**

```python
# Создаём справочник (расшифровку id)
mapping = (
    statistics[["section", "indicator_name", "subsection"]]
    .drop_duplicates()
    .reset_index(drop=True)
    .reset_index()  # index станет id (0, 1, 2, ...)
    .rename(columns={"index": "id"})
)

# Объединяем исходный df со справочником, чтобы получить id
statistics = statistics.merge(
    mapping, on=["section", "indicator_name", "subsection"], how="left"
)

# Оставляем только нужные столбцы
statistics = statistics[["id", "object_name", "indicator_value"]]

# Выводим результат и справочник
print("Результат:")
display(statistics.head())

print("\nСправочник расшифровки id:")
display(mapping.head())
```

```python
glossary = pd.read_excel("sources/glossary.xlsx")
```

Добавляю название файла 

```python
statistics = statistics.merge(
    glossary,
    left_on="object_name",
    right_on="region_source",
    how="inner", 
)


statistics.to_csv(
    "result/statistics.csv", index=False, encoding="utf-8"
)
```

### <font color="green">Загрузка данных карт по каждому субъекту</font> 

```python
map_by_region = statistics[["object_name", "region_map"]].drop_duplicates()
```

```python
display(map_by_region.head())
```

Создание столбцов с features для каждого региона

```python
# Список для сбора итоговых данных по регионам
region_features = []


for idx, row in map_by_region.iterrows():
    name = row["region_map"]
    object_name = row["object_name"]

    path_geojson = f"sources/for_model/{name}.geojson"
    path_pbf = f"sources/for_model/{name}.pbf"

    boundary, layers, all_layers = read_map(path_geojson, path_pbf)
    grid_gdf, cell_w, cell_h = create_grid_within_boundary(boundary, cell_size_km=1)

    final_stats = aggregate_all_layers_to_grid(all_layers, grid_gdf, boundary)

    # Описания и json с типами
    descriptions = final_stats.apply(
        lambda r: create_cell_description(r, all_layers), axis=1
    )
    final_stats["types_json"] = descriptions.apply(lambda x: x[1])

    if "geometry" in final_stats.columns:
        final_stats["center_lon"] = final_stats["geometry"].apply(
            lambda g: g.centroid.x
        )
        final_stats["center_lat"] = final_stats["geometry"].apply(
            lambda g: g.centroid.y
        )

    count_cols = [f"count_{layer}" for layer in all_layers.keys()]
    final_stats["total_objects"] = final_stats[count_cols].sum(axis=1)

    df_main = final_stats.drop(columns=["geometry"]).copy()
    df_main["types_json"] = df_main["types_json"].apply(safe_json_parse)

    # Агрегация по региону
    total_cells = df_main.shape[0]
    empty_cells = (df_main["total_objects"] == 0).sum()

    # Считаем распределение типов объектов (в процентах по региону)
    type_distribution = {}
    for types_json in df_main["types_json"]:
        if isinstance(types_json, dict):
            for layer_data in types_json.get("layers", {}).values():
                for obj_type, count in layer_data.get("types", {}).items():
                    type_distribution[obj_type] = (
                        type_distribution.get(obj_type, 0) + count
                    )

    # Переводим в проценты
    total_objects_sum = sum(type_distribution.values())
    type_distribution_percent = {
        k: v / total_objects_sum if total_objects_sum > 0 else 0
        for k, v in type_distribution.items()
    }

    # Собираем фичи для региона
    region_row = {
        "region_map": name,
        "object_name": object_name,
        "total_cells": total_cells,
        "empty_cells": empty_cells,
        "empty_cells_share": empty_cells / total_cells if total_cells > 0 else 0,
        "total_objects": total_objects_sum,
        **{f"type_{k}_share": v for k, v in type_distribution_percent.items()},
    }

    region_features.append(region_row)

# Итоговый датафрейм для обучения
region_features_df = pd.DataFrame(region_features)
region_features_df.to_csv(
    "result/region_features_for_model.csv", index=False, encoding="utf-8"
)
```
